```python
#Loading Data

import pandas as pd
sales = pd.DataFrame(
{
    "OrderID": [1001,1002,1003,1004,1005,1006,1007,1008],
    "Date": ["2026-01-01","2026-01-01","2026-01-02","2026-01-02","2026-01-03","2026-01-03","2026-01-04","2026-01-04"],
    "Customer": ["Ali","Sarah","John","Ali","Mike","Sarah","John","Fatima"],
    "Category": ["Laptop","Mouse","Laptop","Keyboard","Monitor","Mouse","Laptop","Monitor"],
    "Amount": [1200,25,1350,80,300,30,1100,350],
    "Payment": ["Credit","Debit","Credit","Debit","Credit","Debit","Credit","Credit"]
}
)
sales.to_csv("store-sales.csv", index=False)
```

```python
sales = pd.read_csv("store-sales.csv")
# pd.read_excel()

sales
```

```python
# Exploring Data
sales.head()
sales.tail()
sales.shape
sales.columns
sales.info()
sales.describe()
```

```python
# Series vs DataFarme

sales["Amount"]  # Series -- one column
sales[["Customer", "Amount"]] #DataFrame --full table/multiple columns

```

```python
# Filtering Data
sales[sales["Amount"] > 500]
sales[sales["Customer"] == "Sarah"]
sales[(sales["Amount"] >= 1200) & (sales["Payment"] == "Credit")]
```

```python
# Aggregate Data

sales["Amount"].sum()
sales["Amount"].mean()
sales["Amount"].max()

sales.groupby("Customer")["Amount"].sum()
```

```python
import matplotlib.pyplot as plt

sales_category = sales.groupby("Category")["Amount"].sum()

sales_category.plot(kind="bar")
plt.title("Revenue by product Category")
plt.xlabel("Categories")
plt.ylabel("Revenue")
plt.show()
```
