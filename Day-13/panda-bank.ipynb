```python
import pandas as pd
bank = pd.DataFrame({  "Date": ["2026-01-01","2026-01-02","2026-01-03","2026-01-04","2026-01-05","2026-01-06","2026-01-07","2026-01-08"],
    "Description": ["Costco","Salary","Tim Hortons","Amazon","Gas Station","Walmart","Netflix","Costco"],
    "Category": ["Groceries","Income","Food","Shopping","Transport","Groceries","Subscription","Groceries"],
    "Type": ["Debit","Credit","Debit","Debit","Debit","Debit","Debit","Debit"],
    "Amount": [250,3000,12,120,75,180,18,220] })

bank.to_csv("bank-transactions.csv", index=False)
```

```python
bank = pd.read_csv("bank-transactions.csv")
# Explore Data

bank.head()
bank.tail()
bank.info()
bank.describe()




```

```python
#Filter Transaction
bank[bank["Type"]== "Debit"]

bank[bank["Amount"] > 100]

bank[bank["Category"] == "Groceries"]

# bank[() & ()]

bank[(bank["Amount"] > 100) & (bank["Description"] == "Costco")]




```

```python
# Aggregation
bank.groupby("Category")["Amount"].sum()

bank.groupby("Description")["Amount"].sum()
bank.groupby("Description")["Amount"].sum().sort_values()
bank.groupby("Description")["Amount"].sum().sort_values(ascending=False)


expense = bank[bank["Type"] != "Credit"]
expense.groupby("Description")["Amount"].sum().sort_values(ascending=False)

# One line example
# bank.groupby("Description")["Amount"].sum().loc[lambda x: x < 3000 ].sort_values(ascending=False)



```

```python
import matplotlib.pyplot as plt

bank = pd.read_csv("bank-transactions.csv")

spending = bank[bank["Type"]=="Debit"].groupby("Category")["Amount"].sum()

spending.plot(kind="bar")
plt.title("Spending by Category")
plt.xlabel("Categories")
plt.ylabel("Amount")
plt.show
```
